```python
##practical example

## determine if a year is a leap year nasted condition statement

year =int(input("enter a year"))

if year%4==0:
  if year%100==0:
    if year%400==0:
      print(year,"is a leap year")
    else:
        print(year,"is not a leap year")
  else:
    print(year,"is a leap year")      

else:
  print(year,"is not a leap year")    

   
```

```python
num1 = float(input("Enter a number: "))
num2 = float(input("Enter a number: "))

operators = input("Enter the operator (+,-,*,/): ")

if operators == '+':
    print("Result:", num1 + num2)

else:
    if operators == '-':
        print("Result:", num1 - num2)

    else:
        if operators == '*':
            print("Result:", num1 * num2)

        else:
            if operators == '/':
                print("Result:", num1 / num2)

            else:
                print("You entered an invalid operator")
```
